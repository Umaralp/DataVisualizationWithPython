7) Write a Python program which explains uses of customizing seaborn plots with Aesthetic functions.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns #it is alibrary for data visualization

def sinplot(n=10): #sinplot is used to get sinusoidal curves
    x = np.linspace(0, 14, 100) # (start,stop,intervals)
    for i in range(1, n + 1):
        plt.plot(x, np.sin(x + i * .5) * (n + 2 - i))


sns.set_theme() #provides themes
sns.set_context("notebook", font_scale=1.5, rc={"lines.linewidth": 2.5})

sinplot()
plt.title('Seaborn plots with Aesthetic functions')
plt.show()
```
